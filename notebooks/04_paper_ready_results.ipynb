```python
# Cell 1: Notebook 4 Setup
# Purpose: Load final research results and define paths for paper-ready outputs

import os
import pandas as pd
import numpy as np

PROJECT_ROOT = "/mnt/g/banglafake-detection"

REPORTS_DIR = os.path.join(PROJECT_ROOT, "reports")
FIGURES_DIR = os.path.join(REPORTS_DIR, "figures")

os.makedirs(FIGURES_DIR, exist_ok=True)

print("Notebook 4: Paper-Ready Results & Documentation")
print("Project root:", PROJECT_ROOT)
print("Reports directory:", REPORTS_DIR)
print("Figures directory:", FIGURES_DIR)
```

```python
# Cell 2: Load Final Test Results
# Purpose: Load the final model evaluation metrics for paper-ready reporting

metrics_path = os.path.join(
    REPORTS_DIR,
    "final_test_metrics.csv"
)

metrics_df = pd.read_csv(metrics_path)

display(metrics_df)

print("\nFinal Test Metrics Loaded Successfully.")
```

```python
# Cell 3: Create Paper-Ready Performance Table
# Purpose: Format final test metrics into a clean table for the research paper

paper_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score",
        "ROC-AUC"
    ],
    "Score": [
        metrics_df.loc[0, "Accuracy"],
        metrics_df.loc[0, "Precision"],
        metrics_df.loc[0, "Recall"],
        metrics_df.loc[0, "F1 Score"],
        metrics_df.loc[0, "ROC-AUC"]
    ]
})

paper_results["Score"] = paper_results["Score"].round(4)

display(paper_results)

output_path = os.path.join(
    REPORTS_DIR,
    "paper_ready_performance_table.csv"
)

paper_results.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
# Cell 4: Prepare Confusion Matrix for Paper
# Purpose: Create a clean confusion matrix table from the final test results

cm_path = os.path.join(
    REPORTS_DIR,
    "confusion_matrix.csv"
)

cm_df = pd.read_csv(cm_path, index_col=0)

print("Confusion Matrix:")
display(cm_df)

print("\nTotal Test Samples:", cm_df.values.sum())
print("Total Correct:", np.trace(cm_df.values))
print("Total Errors:", cm_df.values.sum() - np.trace(cm_df.values))
```

```python
# ============================================================
# Notebook: 04_paper_ready_results.ipynb
# Cell 5: Load Ablation, McNemar, ANOVA, Multi-Seed Results
# Purpose: Consolidate all Path A negative-result evidence
# ============================================================

ablation_df = pd.read_csv(os.path.join(REPORTS_DIR, "ablation_comparison_table.csv"), index_col=0)
mcnemar_df = pd.read_csv(os.path.join(REPORTS_DIR, "mcnemar_test_results.csv"))
multi_seed_summary = pd.read_csv(os.path.join(REPORTS_DIR, "multi_seed_summary.csv"), header=[0,1], index_col=0)
anova_df = pd.read_csv(os.path.join(REPORTS_DIR, "anova_architecture_effect.csv"))

print("Ablation (single-seed=42) table:")
display(ablation_df)

print("\nMcNemar (seed=42) results:")
display(mcnemar_df)

print("\nMulti-seed summary (3 seeds):")
display(multi_seed_summary)

print("\nRepeated-Measures ANOVA:")
display(anova_df)
```

```python
# ============================================================
# Cell 6: Build Final Paper-Ready Ablation Table
# Purpose: Single clean table combining mean±std across seeds
#          for the paper's main results section
# ============================================================

metrics = ["Accuracy", "Precision", "Recall", "F1", "ROC-AUC"]
paper_ablation_table = pd.DataFrame(index=multi_seed_summary.index)

for m in metrics:
    mean_col = multi_seed_summary[(m, "mean")]
    std_col = multi_seed_summary[(m, "std")]
    paper_ablation_table[m] = mean_col.round(4).astype(str) + " ± " + std_col.round(4).astype(str)

display(paper_ablation_table)

output_path = os.path.join(REPORTS_DIR, "paper_ready_ablation_table.csv")
paper_ablation_table.to_csv(output_path)
print("\nSaved:", output_path)
```

```python
# ============================================================
# Cell 7: Final Statistical Summary Table (for paper's 
#          Results/Discussion section)
# Purpose: One-glance summary of significance testing
# ============================================================

stat_summary = pd.DataFrame({
    "Test": [
        "McNemar (seed=42): Baseline vs CNN",
        "McNemar (seed=42): Baseline vs Attention",
        "McNemar (seed=42): Baseline vs Full",
        "Repeated-Measures ANOVA (3 seeds): Accuracy",
        "Repeated-Measures ANOVA (3 seeds): F1"
    ],
    "Statistic": [
        mcnemar_df.loc[0, "statistic"],
        mcnemar_df.loc[1, "statistic"],
        mcnemar_df.loc[2, "statistic"],
        anova_df.loc[0, "F_statistic"],
        anova_df.loc[1, "F_statistic"]
    ],
    "p_value": [
        mcnemar_df.loc[0, "p_value"],
        mcnemar_df.loc[1, "p_value"],
        mcnemar_df.loc[2, "p_value"],
        anova_df.loc[0, "p_value"],
        anova_df.loc[1, "p_value"]
    ],
    "Conclusion": [
        "CNN significantly worse than baseline (single seed)",
        "No significant difference",
        "No significant difference (borderline)",
        "No significant architecture effect across seeds",
        "No significant architecture effect across seeds"
    ]
})

display(stat_summary)
stat_summary.to_csv(os.path.join(REPORTS_DIR, "final_statistical_summary.csv"), index=False)
print("\nSaved: final_statistical_summary.csv")
print("\n=== Notebook 4 Path-A evidence consolidation complete ===")
```
