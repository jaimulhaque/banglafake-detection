```python
df = pd.concat(
    [real_df, fake_df],
    ignore_index=True
)

print("Combined dataset shape:", df.shape)
```

```python
print(df["Label"].value_counts())

print("\nPercentage:")
print(
    df["Label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
duplicate_count = df.duplicated(
    subset=["Content"]
).sum()

print("Duplicate content rows:", duplicate_count)
```

```python
duplicate_rows = df[
    df.duplicated(subset=["Content"], keep=False)
].sort_values("Content")

print("Total rows involved in duplicate groups:", len(duplicate_rows))

duplicate_rows[
    ["Article ID", "Headline", "Label", "Content"]
]
```

```python
duplicate_label_check = (
    duplicate_rows
    .groupby("Content")["Label"]
    .nunique()
)

conflicting_duplicates = duplicate_label_check[
    duplicate_label_check > 1
]

print(
    "Duplicate content groups with conflicting labels:",
    len(conflicting_duplicates)
)
```
