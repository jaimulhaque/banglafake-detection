```python
empty_headline = (df["Headline"].fillna("").str.strip() == "").sum()
empty_content = (df["Content"].fillna("").str.strip() == "").sum()

print("Empty headlines:", empty_headline)
print("Empty contents:", empty_content)
```

```python
def clean_bangla_text(text):
    if pd.isna(text):
        return ""
    
    text = str(text).strip()
    
    # Remove HTML tags
    text = re.sub(r"<[^>]+>", " ", text)
    
    # Remove zero-width characters
    text = re.sub(r"[\u200b\u200c\u200d\ufeff]", "", text)
    
    # Normalize whitespace
    text = re.sub(r"\s+", " ", text)
    
    return text.strip()

print("Cleaning function created successfully!")
```

```python
df["Headline"] = df["Headline"].fillna("")
df["Content"] = df["Content"].fillna("")

df["clean_headline"] = df["Headline"].apply(clean_bangla_text)
df["clean_content"] = df["Content"].apply(clean_bangla_text)

print("Headline and content cleaning completed!")
```

```python
df["text"] = (
    df["clean_headline"] + " " + df["clean_content"]
).str.strip()

print("Combined text created successfully!")
```

```python
# ============================================================
# Cell 34: Final Data Quality Check
# Purpose: Verify dataset before train/validation/test split
# ============================================================

print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df[["text", "Label"]].isnull().sum())

print("\nDuplicate texts:")
print(df["text"].duplicated().sum())

print("\nLabel distribution:")
print(df["Label"].value_counts())

print("\nUnique labels:")
print(sorted(df["Label"].unique()))
```

```python
# ============================================================
# Cell 35: Select Model Columns
# Purpose: Create the final dataframe for model training
# ============================================================

model_df = df[[
    "Global_ID",
    "text",
    "Label"
]].copy()

print("Model dataset shape:", model_df.shape)

model_df.head()
```

```python
# ============================================================
# Cell 36: Standardize Label Column
# Purpose: Rename Label to lowercase for easier use in PyTorch
# ============================================================

model_df = model_df.rename(
    columns={"Label": "label"}
)

print(model_df.columns.tolist())
```

```python
# ============================================================
# Cell 37: Create Train + Test Split
# Purpose: Reserve 15% data for final evaluation
# ============================================================

from sklearn.model_selection import train_test_split

train_val_df, test_df = train_test_split(
    model_df,
    test_size=0.15,
    random_state=42,
    stratify=model_df["label"]
)

print("Train + Validation:", len(train_val_df))
print("Test:", len(test_df))
```

```python
# ============================================================
# Cell 38: Create Train + Validation Split
# Purpose: Split remaining data into 70% train and 15% validation
# ============================================================

train_df, validation_df = train_test_split(
    train_val_df,
    test_size=(0.15 / 0.85),
    random_state=42,
    stratify=train_val_df["label"]
)

print("Train:", len(train_df))
print("Validation:", len(validation_df))
print("Test:", len(test_df))
```

```python
# ============================================================
# Cell 39: Verify Class Distribution
# Purpose: Confirm Fake/Real balance is preserved
# ============================================================

print("TRAIN")
print(train_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\nVALIDATION")
print(validation_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\nTEST")
print(test_df["label"].value_counts(normalize=True).mul(100).round(2))
```

```python
# ============================================================
# Cell 40: Data Leakage Check
# Purpose: Ensure no identical text appears across splits
# ============================================================

train_texts = set(train_df["text"])
validation_texts = set(validation_df["text"])
test_texts = set(test_df["text"])

print("Train ∩ Validation:",
      len(train_texts & validation_texts))

print("Train ∩ Test:",
      len(train_texts & test_texts))

print("Validation ∩ Test:",
      len(validation_texts & test_texts))
```

```python
# ============================================================
# Cell 41: Save Processed Datasets
# Purpose: Save final datasets for Notebook 2 and Notebook 3
# ============================================================

train_path = os.path.join(
    PROCESSED_DIR,
    "train.csv"
)

validation_path = os.path.join(
    PROCESSED_DIR,
    "validation.csv"
)

test_path = os.path.join(
    PROCESSED_DIR,
    "test.csv"
)

train_df.to_csv(
    train_path,
    index=False,
    encoding="utf-8-sig"
)

validation_df.to_csv(
    validation_path,
    index=False,
    encoding="utf-8-sig"
)

test_df.to_csv(
    test_path,
    index=False,
    encoding="utf-8-sig"
)

print("✅ Train saved:", train_path)
print("✅ Validation saved:", validation_path)
print("✅ Test saved:", test_path)
```
